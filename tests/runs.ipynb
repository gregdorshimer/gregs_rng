```python
from gregs_rng import lcg
import matplotlib.pyplot as plt
from math import factorial

# Tests of independence of pseudo-random numbers generated by gregs_rng.lcg()

# Hypothesis Test for independence using number of runs up and down

# H0: The numbers are independent.
# HA: The numbers are not not independent.
```

```python
# create random number generator and generate sequence of n = 10000 numbers:
g = lcg(42)
n = 10000
rvs = [next(g) for i in range(n)]
```

```python
# count the number of runs of each size and the total number of runs:
runs = dict()
i = 0
while i < n - 1:
    j = i + 1
    increasing = rvs[i] <= rvs[j]
    if increasing:
        while j < len(rvs) - 1 and rvs[j] <= rvs[j + 1]:
            j += 1
    else:
        while j < len(rvs) - 1 and rvs[j] > rvs[j + 1]:
            j += 1
    if j != i:
        runs[j - i] = runs.get(j - i, 0) + 1
    i = j

num_runs = sum(runs.values())
print(f'num_runs: {num_runs}')
print(f'runs: {runs}')
```

```python
# n = 10,000
# num_runs = 6644

# Under the null hypothesis, expected number of runs and variance of the number of runs is given below:
# E(num_runs) = (2 * n - 1) / 3
# Var(num_runs) = (16 * n - 29) / 90
e_num_runs = (2 * n - 1) / 3
print(f'e_num_runs: {e_num_runs}')
var_num_runs = (16 * n - 29) / 90
print(f'var_num_runs: {var_num_runs}')

# for large n (n >= 30), the test statistic is given below:
# ts = num_runs - E(num_runs) / sqrt(var_num_runs)
ts1 = (num_runs - e_num_runs) / (var_num_runs ** .5)
print(f'ts1: {ts1}')

# critival value is +/- 1.96 for alpha = 0.95 confidence.
cv1 = 1.96
print(f'cv: +/- {cv1}')

# conclusion: ts falls within the 95% confidence interval (-cv < ts < cv) therefore fail to reject hte null hypothesis.
# Independence cannot be rejected on the basis of this test.
```

```python
# Hypothesis Test for independence using length of runs up and down

# H0: The numbers are independent.
# HA: The numbers are not not independent.

# n = 10,000
# num_runs = 6644

# Under the null hypothesis, expected number of runs of length i (Yi) for i from 1 to n-1 is given below:
# E(Yi) = (2 / (i + 3)!) / [n * (i^2 + 3i + 1) - (i^3 + 3i^2 - i - 4)]
# calculate the E(Yi)s:
# eyis is a dict containing i values as keys and E(Yi) for each i as values (mimicking the structure of 'runs')
eyis = dict() 
for i in runs.keys():
    eyis[i] = (2 / factorial(i + 3)) * ((n * ((i ** 2) + (3 * i) + 1)) - ((i ** 3) + (3 * (i ** 2)) - i - 4))
print(f'runs: {runs}')
print(f'eyis: {eyis}')

# the chi-square test statistic is given by the sum across all i's of:
# (Oi - E(Yi))^2 / E(Yi)
# where Oi is the observed number of runs of the given length i
ts = 0
for i in eyis.keys():
    ts += (runs[i] - eyis[i]) ** 2 / eyis[i]
print(f'ts: {ts}')

# the critical value at confidence level alpha = 0.05 is given by the 95th quantile of the related chi-square distribution with k=5 degrees of freedom
cv2 = 11.07
print(f'cv: {cv2}')

# Conclusion: ts2 < cv2, therefore fail to reject the null hypothesis.
# Indepence cannot be rejected on the basis of this test.
```
