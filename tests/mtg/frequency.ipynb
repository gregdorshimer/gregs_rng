```python
from gregs_rng import mtg
import matplotlib.pyplot as plt

# Frequency test of pseudo-random numbers generated by gregs_rng.mtg()

# Chi-square Hypothesis Test for Uniformity

# H0: The numbers are uniformly distributed.
# HA: The numbers are not uniformyl distributed.
```

```python
# create random number generator and generate sequence of n = 10000 numbers:
g = mtg(42)
n = 10000
rvs = [next(g) for i in range(n)]
```

```python
# plot the sequence histogram using 20 bins:
b = 20
plt.hist(rvs, bins = b)
plt.show()
```

```python
# calculate expected number of rvs in each bucket:
ex = n / b
ex
```

```python
# calculate observed number of rvs in each bucket and compile into a list:
counts = [0] * b
for rv in rvs:
    for bucket in range(b):
        if rv <= (bucket + 1) / b:
            counts[bucket] += 1
            break
print(counts)
```

```python
# calculate chi-square test statistic:
diffs = [count - ex for count in counts]
# print(diffs)
squares = [diff * diff for diff in diffs]
# print(squares)
chi_ts = sum([square / ex for square in squares])
print(f'chi_ts: {chi_ts}')
# critical value for chi-square distribution with 19 degrees of freedom and alpha = 0.05: 30.144 (https://www.chisquaretable.net/)
chi_cv = 30.144
print(f'chi_cv: {chi_cv}')

# conclusion: chi_ts < chi_cv, therefore fail to reject H0
```
